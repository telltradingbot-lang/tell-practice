## 2. Look at the data / データを見る

```python
import numpy as np, matplotlib.pyplot as plt
from tell import load_practice, run, Order, PRODUCTS

scenarios = load_practice()
sc = scenarios[0]
fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)
for ax, p in zip(axes, PRODUCTS):
    ax.plot(sc.mid[p], lw=0.8)
    ax.set_ylabel(p)
plt.tight_layout(); plt.show()
```

## 3. Write your bot / ボットを書く

```python
class MyBot:
    def reset(self):
        pass

    def on_tick(self, state):
        # Your logic here / ここにロジックを書く
        if state.t == 0:
            return [Order("BRCK", 5)]
        return []
```

## 4. Score it / スコアを出す

```python
pnls = [run(MyBot(), s)["pnl"] for s in scenarios]
x = np.array(pnls)
print(f"Average P&L {x.mean():.0f}   Win rate {(x > 0).mean():.0%}   ({len(x)} scenarios)")
```
